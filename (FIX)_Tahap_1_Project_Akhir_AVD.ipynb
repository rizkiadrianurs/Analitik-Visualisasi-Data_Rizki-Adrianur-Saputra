## **Outliers Values**

```python
num_cols = ['bmi', 'hemoglobin', 'vitamin_intake', 'iron_intake',
            'iron_deficiency_score', 'total_nutrient_score', 'compliance_rate']
```

```python
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
axes = axes.flatten()
for i, col in enumerate(num_cols):
    # Ensure the column is numeric before plotting, coercing errors and dropping NaNs
    numeric_series_for_boxplot = pd.to_numeric(df[col], errors='coerce').dropna()
    if not numeric_series_for_boxplot.empty:
        axes[i].boxplot(numeric_series_for_boxplot, vert=True)
        axes[i].set_title(col)
    else:
        axes[i].set_title(f"{col} (No numeric data)")
        # Hide axis if no valid numeric data to plot
        axes[i].set_xticks([])
        axes[i].set_yticks([])

if len(num_cols) < len(axes):
    for j in range(len(num_cols), len(axes)): # Hide any remaining empty subplots
        axes[j].set_visible(False)
plt.suptitle('Boxplot Sebelum Capping Outlier', fontsize=13)
plt.tight_layout()
plt.show()
```

Output box plot menunjukkan distribusi data untuk setiap kolom numerik. Terlihat bahwa sebagian besar kolom memiliki titik-titik di luar "kumis" (whiskers) box plot, yang mengindikasikan adanya outlier. Kolom seperti `bmi`, `hemoglobin`, `iron_deficiency_score`, dan `compliance_rate` khususnya menunjukkan banyak outlier baik di sisi atas maupun bawah, yang memerlukan penanganan lebih lanjut.

```python
num_cols = ['bmi', 'hemoglobin', 'vitamin_intake', 'iron_intake',
            'iron_deficiency_score', 'total_nutrient_score', 'compliance_rate']

for col in num_cols:
    # Convert column to numeric, coercing errors, and drop NaNs for accurate quantile calculation
    numeric_series = pd.to_numeric(df[col], errors='coerce').dropna()

    if not numeric_series.empty:
        q1 = numeric_series.quantile(0.25)
        q3 = numeric_series.quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        outlier_count = ((numeric_series < lower) | (numeric_series > upper)).sum()
        outlier_percentage = (outlier_count / len(numeric_series)) * 100
        print(f"{col}: Q1={q1:.2f}, Q3={q3:.2f}, Lower={lower:.2f}, Upper={upper:.2f}, Outliers Percentage={outlier_percentage:.2f}%")
    else:
        print(f"Skipping '{col}' for outlier calculation as it contains no valid numeric data after coercion.")
```

Output Diatas menyajikan Q1, Q3, batas bawah, batas atas, dan persentase outlier untuk setiap kolom numerik. `bmi` dan `hemoglobin` memiliki persentase outlier tertinggi (masing-masing 9.28% dan 8.88%). Kolom lain seperti `vitamin_intake`, `iron_intake`, `iron_deficiency_score`, `total_nutrient_score`, dan `compliance_rate` juga memiliki outlier, meskipun dengan persentase yang lebih rendah. Informasi ini sangat penting untuk memutuskan strategi penanganan outlier, seperti capping atau penghapusan.

# **Data Preparation**

## **Incosistent Values**



Pada tahap ini, dilakukan identifikasi dan koreksi terhadap nilai-nilai yang tidak konsisten pada kolom kategorikal. Ditemukan dua kolom yang memiliki inkonsistensi:

1. **gender**: Terdapat variasi penulisan seperti `'male'`, `'M'`, dan `'F'` yang seharusnya distandarisasi menjadi `'Male'` dan `'Female'`.
2. **iron_intake**: Terdapat nilai `'15mg'` (bertipe string dengan satuan) yang seharusnya berupa angka numerik `15`, dan kolom ini perlu dikonversi ke tipe `float`.
3. **age**: Terdapat nilai non-numerik seperti `'twenty'` yang perlu dikonversi ke numerik.

```python
df['gender'] = df['gender'].replace({'male': 'Male', 'M': 'Male', 'F': 'Female'})
print("Nilai unik gender:")
print(df['gender'].unique())
```

```python
df['iron_intake'] = df['iron_intake'].replace('15mg', '15')
df['iron_intake'] = pd.to_numeric(df['iron_intake'], errors='coerce')
print("Tipe data iron_intake :", df['iron_intake'].dtype)
```

```python
print("Contoh nilai tidak valid pada age:", df[df['age'].apply(lambda x: not str(x).isdigit())]['age'].unique()[:5])
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['age'] = df['age'].fillna(df['age'].median()).astype(int)
print("Tipe data age setelah perbaikan:", df['age'].dtype)
```

Setelah dilakukan pembersihan, kolom `gender` kini hanya memiliki dua nilai yaitu `'Male'` dan `'Female'`.
Kolom `iron_intake` telah dikonversi ke tipe numerik float64,
 dan kolom `age` telah dibersihkan dari nilai non-numerik seperti `'twenty'` dengan imputasi median.

## **Duplicated Values**

Tahap ini melakukan identifikasi dan penghapusan data duplikat. Duplikasi dideteksi berdasarkan kolom `user_id` karena setiap pengguna seharusnya memiliki identitas unik dalam dataset.

Setelah penghapusan baris duplikat, dataset kini hanya berisi baris unik yang siap diproses lebih lanjut.

## **Missing Values**

Pada tahap ini dilakukan penanganan nilai hilang (*missing values*) menggunakan pendekatan **Machine Learning (Iterative Imputer)** untuk kolom numerik agar distribusi data tetap terjaga, sedangkan kolom kategorikal tetap menggunakan **Modus**.

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

num_cols_to_impute = ['bmi', 'hemoglobin', 'iron_intake']

it_imputer = IterativeImputer(random_state=42)

df[num_cols_to_impute] = it_imputer.fit_transform(df[num_cols_to_impute])

print("Imputasi berbasis Machine Learning (IterativeImputer) selesai untuk kolom:", num_cols_to_impute)
```

**Penjelasan Imputasi dengan Machine Learning (Iterative Imputer):**

Berbeda dengan imputasi sederhana (Mean/Median) yang hanya melihat satu kolom saja, `IterativeImputer` menggunakan pendekatan regresi untuk memprediksi nilai yang hilang.

1. **Akurasi Lebih Tinggi**: Algoritma ini mempertimbangkan hubungan (korelasi) antar fitur. Misalnya, nilai `hemoglobin` dapat diprediksi lebih akurat dengan melihat pola pada `iron_intake` dan `bmi` secara bersamaan.
2. **Preservasi Distribusi**: Metode ini membantu menjaga varians dan distribusi data asli, yang sangat penting agar hasil analisis statistik dan model prediksi nantinya tidak bias.
3. **Multivariat**: Teknik ini (juga dikenal sebagai MICE - *Multivariate Imputation by Chained Equations*) secara berulang memperbaiki estimasi nilai yang hilang hingga mencapai titik yang paling logis secara matematis.

```python
for col in ['supplement_brand', 'check_date']:
    mode_val = df[col].mode()[0]
    df[col] = df[col].fillna(mode_val)
    print(f"[{col}] diimputasi dengan modus = {mode_val}")
```

```python
print("Missing values setelah imputasi:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print("\nTotal missing values:", df.isnull().sum().sum())
```

Setelah proses imputasi, tidak terdapat lagi nilai hilang pada dataset.

## **Outliers Values**

```python
def cap_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return series.clip(lower, upper)

for col in num_cols:
    df[col] = cap_outliers(df[col])

print("Capping outlier selesai. Shape dataset:", df.shape)
```

```python
df['check_date'] = pd.to_datetime(df['check_date'], errors='coerce')
df['check_month'] = df['check_date'].dt.month
month_dist = df['check_month'].value_counts().sort_index().reset_index()
month_dist.columns = ['Month', 'Total Checkups']
display(month_dist)
```

Status gizi masyarakat sering kali dipengaruhi oleh faktor musiman, seperti ketersediaan pangan lokal atau perubahan gaya hidup saat hari raya. Dengan mengekstrak `check_month` kita dapat melakukan analisis deret waktu (*time-series*) untuk mendeteksi apakah terdapat bulan-bulan tertentu di mana risiko anemia melonjak secara nasional, sehingga Biovita dapat memberikan peringatan dini (*early warning*) kepada pengguna.
